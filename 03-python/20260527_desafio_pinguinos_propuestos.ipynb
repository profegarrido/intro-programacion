<p>
El dataset palmer penguins es un conjunto de datos sobre pingüinos de tres especies diferentes que anidan en una colonia de Archipiélago Palmer, en la Antártida. Contiene información sobre las medidas corporales de los pingüinos, incluyendo la longitud y profundidad del pico, la longitud de la aleta y la masa corporal, así como información sobre el sexo y la edad de los pingüinos. Este conjunto de datos es ampliamente utilizado en la educación y la investigación para enseñar y ejemplificar técnicas de análisis de datos.</p>
<p>
Hay tres especies de pingüinos en el conjunto de datos: Adelie, Gentoo, y Chinstrap. Conocidos cientificamente como Pygoscelis adeliae, Pygoscelis papua, y Pygoscelis antarctica, respectivamente. Cada especie se identifica en el dataset por su nombre común.</p>
